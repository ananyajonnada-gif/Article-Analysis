```python
import plotly.graph_objects as go

years = sorted(data['last_updated_year'].unique())

fig = go.Figure()

colors = [
    "#00F5FF",  # Cyan
    "#7B68EE",  # Purple
    "#00FF7F",  # Spring Green
    "#FF4D6D",  # Pink
    "#FFD60A",  # Yellow
    "#38BDF8"   # Blue
]

for i, year in enumerate(years):
    temp = (
        data[data['last_updated_year'] == year]
        .clean_tags.value_counts()
        .head(20)
    )

    fig.add_trace(
        go.Bar(
            x=temp.index,
            y=temp.values,
            width=0.55,
            marker=dict(
                color=colors[i % len(colors)],
                line=dict(color="white", width=1)
            ),
            text=temp.values,
            textposition="outside",
            visible=(i == 0)
        )
    )

buttons = []

for i, year in enumerate(years):
    visible = [False] * len(years)
    visible[i] = True

    buttons.append(
        dict(
            label=str(year),
            method="update",
            args=[
                {"visible": visible},
                {"title": f"Top 20 Tags — {year}"}
            ]
        )
    )

fig.update_layout(
    template="plotly_dark",
    title="Top 20 Tags — {}".format(years[0]),
    width=1500,
    height=700,
    bargap=0.35,
    plot_bgcolor="#0b0f19",
    paper_bgcolor="#0b0f19",
    font=dict(color="white", family="Segoe UI"),

    updatemenus=[
        dict(
            buttons=buttons,
            direction="down",
            x=1,
            y=1.18,
            xanchor="right",
            yanchor="top",
            bgcolor="#1e293b",
            bordercolor="#00F5FF",
            borderwidth=1,
            font=dict(color="white")
        )
    ]
)

fig.show()
```

```python
import plotly.express as px

# Count each Broad Category
category_counts = (
    data["Broad Category"]
    .value_counts()
    .reset_index()
)

category_counts.columns = ["Category", "Count"]

# VS Code + Sci-Fi Cyberpunk Navy Palette
colors = [
    "#000814",
    "#001233",
    "#001845",
    "#002855",
    "#003566",
    "#004E89",
    "#00509D",
    "#0066CC",
    "#0077CC",
    "#118AB2",
    "#1D9BF0",
    "#58A6FF"
]

# Create Donut Chart
fig = px.pie(
    category_counts,
    names="Category",
    values="Count",
    hole=0.68,
    color="Category",
    color_discrete_sequence=colors
)

# Styling
fig.update_traces(
    textinfo="none",                    # No labels on chart
    pull=[0.02] * len(category_counts), # Small futuristic separation

    marker=dict(
        line=dict(
            color="#0D1117",
            width=2
        )
    ),

    hovertemplate=
        "<b>%{label}</b><br>"
        "Count: %{value}<br>"
        "Percentage: %{percent}<extra></extra>"
)

# Layout
fig.update_layout(
    template="plotly_dark",

    paper_bgcolor="#0D1117",
    plot_bgcolor="#0D1117",

    width=820,
    height=560,

    title=dict(
        text="<b>Overall Distribution of Broad Categories</b>",
        x=0.5,
        font=dict(
            family="Segoe UI",
            size=22,
            color="#58A6FF"
        )
    ),

    font=dict(
        family="Segoe UI",
        size=13,
        color="white"
    ),

    legend=dict(
        title="",
        orientation="v",
        x=1.02,
        y=0.5,
        bgcolor="rgba(0,0,0,0)",
        borderwidth=0,
        font=dict(
            size=12,
            color="white"
        )
    ),

    margin=dict(
        l=20,
        r=20,
        t=70,
        b=20
    )
)

fig.show()
```
